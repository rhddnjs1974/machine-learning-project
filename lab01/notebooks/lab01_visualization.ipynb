## Plot 1 — Missing values per column (raw data)

```python
# ============ TODO (Plot 1) ============
# Bar chart of lab01.summarize_missing(raw).
# Label both axes and give the figure a title.
# =======================================
s = lab01.summarize_missing(raw)
s.plot(kind="bar")
plt.xlabel("Column")
plt.ylabel("Missing values")
plt.title("Missing values per column (raw data)")
plt.savefig("fig1.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Plot 2 — Distribution of `total_price` (clean data)

```python
# ============ TODO (Plot 2) ============
# Histogram of clean['total_price'] (~30 bins).
# Briefly note in a comment: is it skewed? why?
# =======================================
plt.hist(clean["total_price"], bins=30)
plt.yscale("log")
plt.xlabel("total_price (KRW)")
plt.ylabel("Number of orders (log scale)")
plt.title("Distribution of total_price (clean data)")
plt.savefig("fig2.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Plot 3 — `customer_rating` by category (clean data)

```python
# ============ TODO (Plot 3) ============
# One boxplot per category on a single axis.
# Hint: clean.boxplot(column='customer_rating', by='category')
# =======================================
clean.boxplot(column="customer_rating", by="category")
plt.suptitle("")
plt.title("customer_rating by category (clean data)")
plt.xlabel("Category")
plt.ylabel("Customer rating")
plt.savefig("fig3.png", dpi=150, bbox_inches="tight")
plt.show()
```
